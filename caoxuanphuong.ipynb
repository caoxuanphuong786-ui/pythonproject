```python
#1. tạo dataFram
import pandas as pd
data = {
    "Ho_va_ten": [
        "Nguyễn văn An",
        "Trần văn Bình",
        "lê thị hoa",
        "Phạm minh Tuấn",
        "Võ hoàng Nam",
        "Nguyễn thị lan",
        "Trần minh Đức",
        "Lê quốc Huy",
        "Phạm thị Mai",
        "Hoàng văn Long"
    ],
    "chuyen_can":[9,8,10,7,8,9,6,8,9,5],
    "Giua_ky":[8,7,9,6,8,7,5,7,8,6],
    "Cuoi_ky":[9,8,9,7,7,8,6,9,10,4]
}

df = pd.DataFrame(data)

# Hiển thị bằng bảng dữ liệu
print(df)

```

```python
# 2.Tính điểm tổng kết

df["Tong_ket"] = (
    df["chuyen_can"] * 0.2
    + df["Giua_ky"] * 0.3
    + df["Cuoi_ky"] * 0.5
)

print(df)
```

```python
# 3. Xếp loại sinh viên

def xep_loai(diem):
    if diem >= 8.5:
        return "Giỏi"
    elif diem >= 7.0:
        return "Khá"
    elif diem >= 5.0:
        return "Trung bình"
    else:
        return "Yếu"

df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)

print(df)
```

```python
# 4. Thống kê

# 1. Tính điểm tổng kết trung bình của lớp
diem_trung_binh = df["Tong_ket"].mean()
print("Điểm trung bình của lớp:", round(diem_trung_binh, 2))

# 2. Tìm sinh viên có điểm tổng kết cao nhất
sv_cao_nhat = df.loc[df["Tong_ket"].idxmax()]
print("\nSinh viên có điểm cao nhất:")
print(sv_cao_nhat["Ho_va_ten"], "-", sv_cao_nhat["Tong_ket"])

# 3. Tìm sinh viên có điểm tổng kết thấp nhất
sv_thap_nhat = df.loc[df["Tong_ket"].idxmin()]
print("\nSinh viên có điểm thấp nhất:")
print(sv_thap_nhat["Ho_va_ten"], "-", sv_thap_nhat["Tong_ket"])

# 4. Đếm số sinh viên đạt từ 5 điểm trở lên
so_sv_dat = (df["Tong_ket"] >= 5).sum()
print("\nSố sinh viên đạt:", so_sv_dat)
```

```python
# 5. Trực quan hóa điểm tổng kết

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(df["Ho_va_ten"], df["Tong_ket"])

plt.title("Diem tong ket cua 10 sinh vien")
plt.xlabel("Ho va ten sinh vien")
plt.ylabel("Diem tong ket")

plt.xticks(rotation=45, ha="right")
plt.ylim(0, 10)

plt.tight_layout()
plt.show()
```
